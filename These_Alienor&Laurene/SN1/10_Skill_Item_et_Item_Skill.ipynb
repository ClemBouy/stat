```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from fanalysis.ca import CA
from matplotlib.ticker import PercentFormatter

data = pd.read_csv('data.csv')
```

### Diagrammes en barre décrivant la proportion des item abordés dans chaque compétence. 

Les différentes compétences sont présentées par ordre décroissant. 

```python
data_gp = pd.pivot_table(data, values="record_id", index="skill", columns="ecn_item", aggfunc="count", margins=True)
data_gp = data_gp.sort_values(by=['All'], ascending=False)
data_gp = data_gp.drop(columns='All')
display(data_gp)

for i in range(1,data_gp.shape[0]):
    row = data_gp.iloc[[i]]
    row = row.dropna(axis=1, how='all')
    row = row[row.columns[row.iloc[0].argsort()[::-1]]]

    rowProp = row.copy()
    rowProp[row.columns] = row[row.columns].div(row[row.columns].sum(axis=1), axis=0)

    plt.figure(figsize=(20, 10))
    sns.barplot(rowProp)
    plt.xticks(rotation=90)
    plt.title(f"Diagramme en barres pour la compétence {data_gp.index[i]}")
    plt.gca().yaxis.set_major_formatter(PercentFormatter(1.0))
    plt.show()  
    print("Valeurs absolues : \n")
    display(row)
    print("Proportions : \n")
    display(rowProp)
```

### Diagrammes en barre décrivant la proportion des compétences dans chaque item. 

Les différents item sont présentées par ordre décroissant. 

```python
data_gp = pd.pivot_table(data, values="record_id", index="ecn_item", columns="skill", aggfunc="count", margins=True)
data_gp = data_gp.sort_values(by=['All'], ascending=False)
data_gp = data_gp.drop(columns='All')
display(data_gp)

palette = sns.color_palette("Set3", 6)

color_map = {
    'C1' : palette[0],
    'C2' : palette[1],
    'C3' : palette[2],
    'C4' : palette[3],
    'C5' : palette[4],
    'C6' : palette[5],
}

for i in range(1,data_gp.shape[0]):
    row = data_gp.iloc[[i]]
    row = row.dropna(axis=1, how='all')
    row = row[row.columns[row.iloc[0].argsort()[::-1]]]

    active_cols = row.columns
    colors = [color_map[col] for col in active_cols]

    rowProp = row.copy()
    rowProp[row.columns] = row[row.columns].div(row[row.columns].sum(axis=1), axis=0)

    plt.figure(figsize=(20, 10))
    sns.barplot(rowProp, palette=colors)
    plt.xticks(rotation=90)
    plt.title(f"Diagramme en barres pour le RC {data_gp.index[i]}")
    plt.gca().yaxis.set_major_formatter(PercentFormatter(1.0))
    plt.show()  
    print("Valeurs absolues : \n")
    display(row)
    print("Proportions : \n")
    display(rowProp)
```

### Analyse factorielle des correspondances entre les variables "Item" et "Skill"

```python
# Création et affichage du tableau de contingence. 

data_factor = pd.pivot_table(data, values="record_id", index=[data['ecn_item']], columns=[data['skill']], aggfunc="count", fill_value=0)
display(data_factor)

col_names = list(data_factor.columns)[0:]
row_names = list(data_factor.index)

X = data_factor[col_names].values
```

```python
# Calcul et affichage des profils lignes (la somme de chaque ligne est égale à 1.)

row_sum = np.sum(X, axis=1)
row_profile = X / row_sum[:, np.newaxis]

print("Profil ligne:")
display(pd.DataFrame(row_profile, index=row_names, columns=col_names))
```

```python
# Calcul et affichage des profils colonnes (la somme de chaque colonne est égale à 1).

col_sum = np.sum(X, axis=0)
col_profile = X / col_sum[np.newaxis, :]

print("Profil colonne:")
display(pd.DataFrame(col_profile, index=row_names, columns=col_names))
```

```python
# Initialisation de l'objet CA et estimation du modèle. 
my_ca = CA(row_labels=data_factor.index.values, col_labels=data_factor.columns.values,stats=True)
my_ca.fit(data_factor.values)
```

```python
# Affichage des valeurs propres
my_ca.eig_
```

```python
# Graphique des valeurs propres
my_ca.plot_eigenvalues(type="percentage")
```

```python
# Graphique cumulé des valeurs propres.  
my_ca.plot_eigenvalues(type="cumulative")
```

```python
# Affichage des cos2 pour chaque variable. 

print("Premier axe factoriel")
my_ca.plot_row_cos2(num_axis=1, figsize=(20,30))
print("Second axe factoriel")
my_ca.plot_row_cos2(num_axis=2, figsize=(20,30))

print("Premier axe factoriel")
my_ca.plot_col_cos2(num_axis=1, figsize=(10,10))
print("Second axe factoriel")
my_ca.plot_col_cos2(num_axis=2, figsize=(10,10))
```

```python
# Affichage des contributions pour chaque variable
print("Premier axe")
my_ca.plot_row_contrib(num_axis=1, figsize=(20,30))
print("Second axe")
my_ca.plot_row_contrib(num_axis=2, figsize=(20,30))

print("Premier axe")
my_ca.plot_col_contrib(num_axis=1, figsize=(10,10))
print("Second axe")
my_ca.plot_col_contrib(num_axis=2, figsize=(10,10))
```

```python
# AFC
model = CA(n_components=2)
model.fit(data_factor.values)

cos2_seuil = 0.2
contrib_seuil = 3

# Coordonnées
row_coords = model.row_coord_
col_coords = model.col_coord_

# Définir un niveau de jitter (ajustable)
jitter_strength = 0.1

# cos² pour l'axe 1
row_cos2 = model.row_cos2_[:, 0]
col_cos2 = model.col_cos2_[:, 0]

# Contributions à l'axe 1 (0)
row_contrib = model.row_contrib_[:, 0]
col_contrib = model.col_contrib_[:, 0]

# Tracé
fig, ax = plt.subplots(figsize=(15, 15))

# Modalités lignes (cercles)
for i, label in enumerate(data_factor.index):
    cos2_axe1 = model.row_cos2_[i, 0]
    cos2_axe2 = model.row_cos2_[i, 1]
    contrib_axe1 = model.row_contrib_[i, 0]
    contrib_axe2 = model.row_contrib_[i, 1]
    
    if (cos2_axe1 > cos2_seuil or cos2_axe2 > cos2_seuil or
        contrib_axe1 > contrib_seuil or contrib_axe2 > contrib_seuil):

        color = 'red' if cos2_axe1 > 0.5 else 'blue' if cos2_axe2 > 0.5 else 'gray'

        # Appliquer le jitter
        x = row_coords[i, 0] + np.random.uniform(-jitter_strength, jitter_strength)
        y = row_coords[i, 1] + np.random.uniform(-jitter_strength, jitter_strength)

        ax.scatter(x, y, color=color, marker='x', s=100)
        ax.text(x + 0.02, y + 0.02, label, color=color, fontsize=8)

# Modalités colonnes (triangles)
for i, label in enumerate(data_factor.columns):
    cos2_axe1 = model.col_cos2_[i, 0]
    cos2_axe2 = model.col_cos2_[i, 1]
    contrib_axe1 = model.col_contrib_[i, 0]
    contrib_axe2 = model.col_contrib_[i, 1]
    
    if (cos2_axe1 > cos2_seuil or cos2_axe2 > cos2_seuil or
        contrib_axe1 > contrib_seuil or contrib_axe2 > contrib_seuil):

        # Déterminer la couleur
        if cos2_axe1 > 0.5:
            color = 'red'
        elif cos2_axe2 > 0.5:
            color = 'blue'
        else:
            color = 'gray'

        # Ajouter jitter
        x = model.col_coord_[i, 0] + np.random.uniform(-jitter_strength, jitter_strength)
        y = model.col_coord_[i, 1] + np.random.uniform(-jitter_strength, jitter_strength)

        # Tracer le point et le label
        ax.scatter(x, y, color=color, marker='x', s=100)
        ax.text(x + 0.02, y + 0.02, label, color=color, fontsize=8)


# Axes et style
ax.axhline(0, color='black', linestyle='-', linewidth=1.5)
ax.axvline(0, color='black', linestyle='-', linewidth=1.5)
ax.set_title("AFC - Couleur selon contribution à l'axe 1")
ax.set_xlabel('Axe 1')
ax.set_ylabel('Axe 2')
plt.grid(False)
plt.tight_layout()
plt.show()
```
