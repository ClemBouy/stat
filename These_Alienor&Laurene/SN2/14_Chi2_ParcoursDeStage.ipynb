```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from fanalysis.ca import CA
from matplotlib.ticker import PercentFormatter

from scipy.stats import chi2_contingency  
from scipy.stats import chi2

data = pd.read_csv('data.csv')
```

## Chi2 d'indépendance : famille de situation en fonction du parcours de stage. 

### Pour SN1 et Urgences

```python
data_FS = pd.pivot_table(data=data, values="record_id", columns="order_1", index="FS", aggfunc="count")
display(data_FS)
```

```python
chi_squared_result=chi2_contingency(data_FS)
alpha = 0.05

print("La statistique de Pearson du test khi-deux est : ", chi_squared_result[0], "\n"
      "La p-valeur du test est : ", chi_squared_result[1], "\n"
      "Nombre de degré de liberté : ", chi_squared_result[2], "\n")
print("Le quantile de la loi du chi 2 à", chi_squared_result[2], "degré de liberté associé au niveau de confiance",alpha,"est",chi2.ppf(1-alpha, chi_squared_result[2]) , "\n")

print("Valeurs attendues : \n", chi_squared_result[3])
```

**Conclusion** : La p-value est très supérieure à 0.05 on ne peut pas rejeter l'hypothèse nulle. Il ne semble pas y avoir de corrélation entre le parcours de stage et les familles de situation abordées.  
A noter que certaines valeurs observées et attendues sont inférieures à 5 donc les conditions de validité du test ne sont théoriquement pas vérifiées.
