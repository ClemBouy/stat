```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from fanalysis.ca import CA
from matplotlib.ticker import PercentFormatter

from scipy.stats import chi2_contingency  
from scipy.stats import chi2
from scipy.stats import fisher_exact
from scipy.stats import MonteCarloMethod

data = pd.read_csv('data.csv')
```

```python
# On ne garde que les 3 premiers caractères de la colonne tag pour comparer les SN1 et les SN2 indépendamment de la réalisation de stages supplémentaires ou non.
data['tag'] = data['tag'].str[:3]
```

## Comparaison de la distribution des familles de situation

```python
data_FS = pd.pivot_table(data=data, values="record_id", columns="tag", index="FS", aggfunc="count")
data_FS
```

```python
chi_squared_result=chi2_contingency(data_FS)
alpha = 0.05

print("\033[1;32mTest du Chi2 :\033[0m")
print("La statistique de Pearson du test khi-deux est : ", chi_squared_result[0], "\n"
      "La p-valeur du test est : ", chi_squared_result[1], "\n"
      "Nombre de degré de liberté : ", chi_squared_result[2], "\n")
print("Le quantile de la loi du chi 2 à", chi_squared_result[2], "degré de liberté associé au niveau de confiance",alpha,"est",chi2.ppf(1-alpha, chi_squared_result[2]) , "\n")

print("Valeurs attendues : \n", chi_squared_result[3])
```

```python
col_names = list(data_FS.columns)[0:]
row_names = list(data_FS.index)

X = data_FS[col_names].values

col_sum = np.sum(X, axis=0)
col_profile = (X / col_sum[np.newaxis, :]).round(4)

df = pd.DataFrame(col_profile, index=row_names, columns=col_names)
print("Distribution des familles de situation en SN1 et SN2")
display(df)
```

**Conclusion** : la distribution des familles de situation est significativement différente entre SN1 et SN2 (en analyse univariée).  
En SN1, les 4 FS les plus représentées sont les FS02, 01, 04 et 03. En SN2, on retrouve d'abord les FS02, 01, 08 et 04.  
D'une façon générale, les familles de situation les 4 premières familles de situation, qu'on observe plus fréquemment dans le RAQ (autre thèse) tout niveau de stage confondus semblent moins proposées en SN2 par rapport au SN1. Peut-être par soucis de valoriser des FS qui manqueraient pour certifier ?

## Comparaison de la distribution des items

```python
data_item = pd.pivot_table(data=data, values="record_id", columns="tag", index="ecn_item", aggfunc="count", fill_value=0)
data_item
```

```python
chi_squared_result=chi2_contingency(data_item)
alpha = 0.05

print("\033[1;32mTest du Chi2 :\033[0m")
print("La statistique de Pearson du test khi-deux est : ", chi_squared_result[0], "\n"
      "La p-valeur du test est : ", chi_squared_result[1], "\n"
      "Nombre de degré de liberté : ", chi_squared_result[2], "\n")
print("Le quantile de la loi du chi 2 à", chi_squared_result[2], "degré de liberté associé au niveau de confiance",alpha,"est",chi2.ppf(1-alpha, chi_squared_result[2]) , "\n")

print("Valeurs attendues : \n", chi_squared_result[3])
```

**Conclusion** : le Chi2 est significatif mais le tableau de contingence comporte de nombreuses valeurs nulles ou inférieures à 5. Il n'est donc théoriquement pas possible d'interpréter les résultats du Chi2.  
Pour essayer de lever cette limite, les valeurs inférieurs à 5 vont être rangés dans une catégorie "autre".

```python
# Ajout d'une colonne "total" pour calculer les effectifs ligne par ligne
data_item['total'] = data_item['SN1'] + data_item['SN2']
data_item.reset_index(inplace=True)

# Seuil pour regrouper les lignes faibles (ex. : < 5)
seuil = 5

# Création d'un groupe : on garde le nom de l'item ou on remplace par "Autres"
data_item['groupe'] = data_item.apply(
    lambda row: row['ecn_item'] if row['SN1'] >= seuil and row['SN2'] >= seuil else 'Autres',
    axis=1
)

# Regroupement par groupe et somme des effectifs
df_grouped = data_item.groupby('groupe')[['SN1', 'SN2']].sum().reset_index()

print("\033[1;32mTest du Chi2 :\033[0m")
print(df_grouped)

# Tableau de contingence
contingency = df_grouped[['SN1', 'SN2']].values

df_prop = df_grouped.set_index('groupe')

# Test du Chi²
chi2, p, dof, expected = chi2_contingency(contingency)

# Affichage des résultats
print(f"Chi² = {chi2:.2f}")
print(f"p-value = {p:.4f}")
print(f"Degrés de liberté = {dof}")
print("\nTableau des fréquences attendues :\n", pd.DataFrame(expected, columns=['SN1', 'SN2'], index=df_grouped['groupe']))
```

```python
col_names = list(df_prop.columns)[0:]
row_names = list(df_prop.index)

X = df_prop[col_names].values

col_sum = np.sum(X, axis=0)
col_profile = (X / col_sum[np.newaxis, :]).round(4)

df = pd.DataFrame(col_profile, index=row_names, columns=col_names)
print("Distribution des items en SN1 et SN2")
display(df)
```

**Conclusion:** Après création de la catégorie "Autres" le chi2 reste significatif. La distribution des Item entre SN1 et SN2 est donc bien significativement différente. Le tableau ci-dessus décrit les proportions des différents items.

## Comparaison des distributions pour les 10 premiers items

```python
top10_items = ['Item 1', 'Item 13', 'Item 146', 'Item 19', 'Item 204', 'Item 224', 'Item 247', 'Item 290', 'Item 361', 'Item 66']
```

### Item et problématiques exemplaire

```python
# Pour les problématiques exemplaires 

rng = np.random.default_rng()
method = MonteCarloMethod()

for item in top10_items:
    print(f"\033[1;32mTest exact de Fisher avec méthode de Monte Carlo ({item}):\033[0m")
    sample = data.loc[data['ecn_item'] == item]
    sample = pd.pivot_table(data=sample, values="record_id", columns="tag", index="PE", aggfunc="count", fill_value=0)
    contingency = sample[['SN1', 'SN2']].values
    print(contingency)
    odds_ratio, p_value = fisher_exact(contingency, method=method)
    print(f"\nTableau de contingence pour {item} :")
    print(f"p-value : {p_value}")
    display(sample)

```

### Item et étape du raisonnement

```python
for item in top10_items:
    sample = data.loc[data['ecn_item'] == item]
    sample = sample.groupby(['tag']).agg(
        diagnostic = ('step_diag', 'sum'),
        prescription = ('step_PEC_TTT', 'sum'), 
        autre = ('step_autre', 'sum')
    )

    sample = sample.T

    contingency = sample[['SN1', 'SN2']].values
    print(f"\nTableau de contingence pour {item} :")
    display(sample)
    try:
        odds_ratio, p_value = fisher_exact(contingency, method=method)
        print(f"\033[1;32mTest exact de Fisher avec méthode de Monte Carlo ({item}):\033[0m")
        print(f"p-value : {p_value}")
    except Exception as e: 
        print(f"[ERREUR] : {e}")
    
```
